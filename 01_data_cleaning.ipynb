```python
df["ctr"] = df["clicks"] / df["impressions"]

df[["clicks", "impressions", "ctr"]].head()
```

```python
df["conversion_rate_actual"] = (
    df["conversions"] / df["clicks"]
)

df[["clicks", "conversions", "conversion_rate_actual"]].head()
```

```python
df["conversion_rate_actual"] = (
    df["conversions"] / df["clicks"]
)

df[["clicks", "conversions", "conversion_rate_actual"]].head()
```

```python
df["cpc_actual"] = df["cost"] / df["clicks"]

df[["cost", "clicks", "cpc_actual"]].head()
```

```python
df["cost_per_conversion"] = (
    df["cost"] / df["conversions"]
)

df[["cost", "conversions", "cost_per_conversion"]].head()
```

```python
platform_summary = df.groupby("platform").agg({
    "impressions": "mean",
    "clicks": "mean",
    "cost": "mean",
    "conversions": "mean",
    "ctr": "mean",
    "conversion_rate_actual": "mean",
    "cost_per_conversion": "mean"
})

platform_summary
```

```python
region_summary = df.groupby("region").agg({
    "conversions": "mean",
    "cost_per_conversion": "mean",
    "ctr": "mean"
})

region_summary
```

```python
df["date"] = pd.to_datetime(df["date"])

df.info()
```

```python
df["month"] = df["date"].dt.month

monthly_conversions = df.groupby("month")["conversions"].sum()

monthly_conversions
```
