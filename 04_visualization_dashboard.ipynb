```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("C:\\Users\\pooja\\OneDrive\\Desktop\\Data analyst(Pooja)\\pooja_portfolio_2026\\SMM_3_ABtesting_marketting\\cleaned_marketing_data.csv")

df.head()
```

```python
platform_perf = df.groupby("platform")[["clicks", "conversions", "cost"]].mean()

platform_perf.plot(kind="bar", figsize=(8,5))

plt.title("Platform Performance Comparison")
plt.ylabel("Average Value")
plt.show()
```

```python
🧠 Explanation

Shows:

which platform performs better
engagement vs cost efficiency
```

```python
df["date"] = pd.to_datetime(df["date"])

trend = df.groupby("date")["conversions"].sum()

plt.figure(figsize=(10,5))
plt.plot(trend)

plt.title("Daily Conversion Trend")
plt.xlabel("Date")
plt.ylabel("Conversions")
plt.xticks(rotation=45)
plt.show()
```

```python
🧠 Explanation

This shows:

campaign performance over time
spikes (festivals / weekends)
drops (ad fatigue)
```

```python
df["ctr"] = df["clicks"] / df["impressions"]

plt.figure(figsize=(8,5))
sns.histplot(df["ctr"], bins=30, kde=True)

plt.title("CTR Distribution")
plt.show()
```

```python
🧠 Explanation

Shows:

how users interact with ads
distribution of engagement quality
```

```python
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="cost", y="conversions", hue="platform")

plt.title("Cost vs Conversions")
plt.show()
```

```python
🧠 Explanation

This shows:

ROI efficiency
whether spending more increases conversions
platform efficiency comparison
```

```python
plt.figure(figsize=(10,6))

corr = df[["impressions", "clicks", "cost", "conversions", "ctr"]].corr()

sns.heatmap(corr, annot=True, cmap="coolwarm")

plt.title("Feature Correlation Heatmap")
plt.show()
```

```python
🧠 Explanation

This shows:

which features are strongly related
helps explain ML model behavior
```

```python
region_perf = df.groupby("region")["conversions"].mean()

region_perf.plot(kind="bar")

plt.title("Region-wise Conversion Performance")
plt.ylabel("Average Conversions")
plt.show()
```

```python
🧠 Explanation

Shows:

best performing regions
geo targeting insights
```

```python
hour_perf = df.groupby("hour")["conversions"].mean()

plt.figure(figsize=(10,5))
plt.plot(hour_perf)

plt.title("Hourly Conversion Pattern")
plt.xlabel("Hour of Day")
plt.ylabel("Conversions")
plt.show()
```
