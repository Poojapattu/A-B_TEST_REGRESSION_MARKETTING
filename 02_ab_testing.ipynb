```python
import pandas as pd
import numpy as np
from scipy import stats
```

```python
df = pd.read_csv("C:\\Users\\pooja\\OneDrive\\Desktop\\Data analyst(Pooja)\\pooja_portfolio_2026\\SMM_3_ABtesting_marketting\\cleaned_marketing_data.csv")

df.head()
```

```python
df["cr"] = df["conversions"] / df["clicks"]

df[["platform", "clicks", "conversions", "cr"]].head()
```

```python
google_group = df[df["platform"] == "Google"]["cr"]

instagram_group = df[df["platform"] == "Instagram"]["cr"]

print("Google Samples:", len(google_group))
print("Instagram Samples:", len(instagram_group))
```

```python
print("Google Mean CR:", google_group.mean())
print("Instagram Mean CR:", instagram_group.mean())
```

```python
# Hypothesis Testing

## Null Hypothesis (H₀)
##There is no significant difference in conversion rates between Google and Instagram ads.

## Alternative Hypothesis (H₁)
##There is a significant difference in conversion rates between the two platforms.
```

```python
t_statistic, p_value = stats.ttest_ind(
    google_group,
    instagram_group,
    equal_var=False
)

print("T-Statistic:", t_statistic)
print("P-Value:", p_value)
```

```python
alpha = 0.05

if p_value < alpha:
    print("✅ Significant difference found")
else:
    print("❌ No significant difference")
```

```python
if google_group.mean() > instagram_group.mean():
    print("Google Ads perform better based on conversion rate.")
else:
    print("Instagram Ads perform better based on conversion rate.")
```
